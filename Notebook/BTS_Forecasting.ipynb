# **Electric ARIMA**

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
```

```python
bts_df['Year'] = bts_df['Year'].astype(int)
bts_df = bts_df.sort_values('Year')
```

```python
bts_df.set_index('Year', inplace=True)
```

```python
series = bts_df['Electric']
```

```python
model = ARIMA(series, order=(1,1,1))
model_fit = model.fit()

forecast = model_fit.forecast(steps=7)

future_years = list(range(series.index.max() + 1, series.index.max() + 8))
```

```python
plt.figure()

plt.plot(series.index, series, marker='o', label='Actual')
plt.plot(future_years, forecast, marker='o', linestyle='--', label='Forecast')

plt.title("Electric Vehicle Sales Forecast (ARIMA)")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

# **Electric Linear Regression**

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
bts_df = bts_df.sort_values('Year')
```

```python
bts_df['lag_1'] = bts_df['Electric'].shift(1)
bts_df['lag_2'] = bts_df['Electric'].shift(2)
bts_df['growth'] = bts_df['Electric'].pct_change()

bts_df['future'] = bts_df['Electric'].shift(-1)
```

```python
bts_df = bts_df.dropna()
```

```python
X = bts_df[['lag_1', 'lag_2']]
y = bts_df['future']
```

```python
split = int(len(bts_df) * 0.7)

train = bts_df.iloc[:split]
test = bts_df.iloc[split:]
```

```python
X_train = train[['lag_1','lag_2']]
y_train = train['future']

X_test = test[['lag_1','lag_2']]
y_test = test['future']
```

```python
# Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

print("LR MAE:", mean_absolute_error(y_test, y_pred_lr))
print("LR R2:", r2_score(y_test, y_pred_lr))
```

```python
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

print("RF MAE:", mean_absolute_error(y_test, y_pred_rf))
print("RF R2:", r2_score(y_test, y_pred_rf))
```

# **Electric Forecasting Pipleline**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

bts_df = pd.read_csv('/content/drive/My Drive/Data/Capstone/bts_data.csv')

bts_df['Year'] = bts_df['Year'].astype(int)
bts_df = bts_df[(bts_df['Year'] >= 2010) & (bts_df['Year'] <= 2023)]
bts_df = bts_df.sort_values('Year').reset_index(drop=True)
```

```python
sales_col = 'Electric'

bts_df['lag_1'] = bts_df[sales_col].shift(1)
bts_df['lag_2'] = bts_df[sales_col].shift(2)
bts_df['growth'] = bts_df[sales_col].pct_change()
```

```python
bts_df['future_sales'] = bts_df[sales_col].shift(-1)
```

```python
bts_df = bts_df.dropna(subset=['lag_1', 'lag_2', 'growth', 'future_sales'])
```

```python
features = ['lag_1', 'lag_2', 'growth']

train = bts_df.iloc[:-1]
test = bts_df.iloc[-1:]

X_train = train[features]
y_train = train['future_sales']

X_test = test[features]
y_test = test['future_sales']
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
last_year = bts_df['Year'].max()
future_years = list(range(last_year + 1, 2031))

lag1 = bts_df.iloc[-1][sales_col]
lag2 = bts_df.iloc[-2][sales_col]
growth = bts_df.iloc[-1]['growth']
```

```python
predictions = []

for year in future_years:

    X_future = pd.DataFrame({
        'lag_1': [lag1],
        'lag_2': [lag2],
        'growth': [growth]
    })

    pred = model.predict(X_future)[0]
    predictions.append(pred)

    lag2 = lag1
    lag1 = pred

forecast_df = pd.DataFrame({
    'Year': future_years,
    'Predicted_Electric': predictions
})
```

```python
plt.figure()

plt.plot(bts_df['Year'], bts_df[sales_col],
         marker='o', label='Actual Electric')

plt.plot(forecast_df['Year'], forecast_df['Predicted_Electric'],
         marker='o', linestyle='--', label='Forecast Electric')

plt.title("Electric Vehicle Sales Forecast (2010–2030)")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

# **Results**

Linear Regression

Result: Very low or negative R² values. Not Good.

Findings:
- Model struggled to accurately predict EV demand
- Indicates that EV growth is not linear
- Shows that a simple straight-line model is insufficient

Random Forest Results

Result: Worse performance than linear regression (very negative R²)

Findings:
- Model overfit due to small dataset
- Not good for time-series forecasting with this data

ARIMA Model Results

Result: Smooth, realistic forecasts with consistent upward trend

Findings:
- Best-performing model for this dataset
- Effectively shows time-based trends and growth patterns
- Provides reliable projections through 2030

Electric Vehicle (EV) Trend

Result: Strong and accelerating growth over time

Findings:
- EV demand is increasing rapidly
- Indicates a long-term shift toward full electrification
- Represents the highest growth opportunity

Hybrid & Plug-in Hybrid Trends

Result: Moderate, steady growth

Findings:
- Demand is increasing but at a slower rate than EVs
- These technologies act as transitional solutions
